We use OpenAI's Whisper to turn the original voice into English, if the translation is meant to be in English. This helps us understand and process speech in other languages.

```python
client = OpenAI()
audio_file = open("audio/French.mp3", "rb")

translation = client.audio.translations.create(
    model="whisper-1", 
    file=audio_file,
)
english_text = translation.text
```

(optional): storing english_text in a file named as "translate.txt"

```python
with open("translate.txt", "w", encoding="utf-8") as file:
    file.write(english_text)
print(english_text)
```

Now using OpenAI Chat Completion to translate it into the preferred language (e.g., Hindi).

```python
preferred_language = 'Hindi'
completion = client.chat.completions.create(
    model="gpt-4o",
    messages=[
        {"role": "system", "content": "You are a translator."},
        {"role": "user", "content": f"Translate this English text to {preferred_language}:\n{english_text}"}
    ]
)

hindi_text = completion.choices[0].message.content
```

Now use hindi HeyGen for Video & Lip-syn

```python
HEY_GEN_KEY = os.getenv("HEYGEN_API_KEY")
```

created url for voice message which heygen will be used in generated video

```python
import requests

url = "https://upload.heygen.com/v1/asset"
headers = {
    "Content-Type": "audio/mpeg",
    "x-api-key": HEY_GEN_KEY
}

with open("audio/hindi_voice.mp3", "rb") as f:
    response = requests.post(url, data=f, headers=headers)

print(response.status_code)
print(response.json())
```

Generate video id

```python
url = "https://api.heygen.com/v1/video.webm"

payload = {
    "avatar_pose_id": "Vanessa-invest-20220722",
    "avatar_style": "normal",
    "dimension": {
        "width": 1280,
        "height": 720
    },
    "input_audio": "https://resource2.heygen.ai/audio/ea949ef2228c41aca7580d376f485575/original"
}
headers = {
    "accept": "application/json",
    "content-type": "application/json",
    "x-api-key": HEY_GEN_KEY
}

response = requests.post(url, json=payload, headers=headers)

print(response.text)
```
